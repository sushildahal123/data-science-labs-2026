```python
import numpy as np
np.random.seed(42) 
```

```python
"""problem-1:You work for a weather-data startup expanding into the US market. Convert an array of Celsius temperatures to Fahrenheit using the
formula F = C * 9/5 + 32 ."""
def celsius_to_fahrenheit(celsius_array):
        """
        celsius_array: 1D NumPy array of temperatures in Celsius
        returns: 1D NumPy array of temperatures in Fahrenheit
        """
            
    return celsius_array * 9/5 + 32

# Test
test_c = np.array([0, 20, 37, 100])
print(celsius_to_fahrenheit(test_c)) 
```

```python
"""problem-2: A fintech app logs transaction amounts (in KRW). Return only the transactions greater than or equal to 50,000 KRW."""
def filter_high_value(transactions, threshold=50000):
    """
    transactions: 1D NumPy array of transaction amounts
    threshold: minimum amount to keep
    returns: 1D NumPy array containing only transactions >= threshold
    """
    return transactions[transactions >= threshold]

# Test
test_tx = np.array([12000, 55000, 49999, 100000, 50000])
print(filter_high_value(test_tx))  # expected: [ 55000 100000  50000]
```

```python
"""problem-3: An IoT sensor logs one temperature reading every hour for 7 days (168 readings total), stored as a flat 1D array. Reshape it into a (7, 24)
array so each row represents one day."""

def reshape_to_days(readings):
    """
    readings: 1D NumPy array of length 168
    returns: 2D NumPy array of shape (7, 24)
    """
    return readings.reshape(7, 24)

# Test
test_readings = np.arange(168)
result = reshape_to_days(test_readings)
print(result.shape) 
print("day 1 = ",result[0])
```

```python
"""problem-4: Using the reshaped (7, 24) sensor array from Problem 3, find the minimum and maximum temperature for each day in a single
vectorized call each (no loop over days)."""

def daily_min_max(daily_readings):
    """
    daily_readings: 2D NumPy array of shape (n_days, 24)
    returns: tuple of two 1D arrays (daily_min, daily_max), each of length n_days
    """
    daily_min = daily_readings.min(axis=1)
    daily_max = daily_readings.max(axis=1)
    return daily_min, daily_max

# Test
test_data = reshape_to_days(np.arange(168))
mins, maxs = daily_min_max(test_data)
print("daily mins:", mins)  
print("daily maxs:", maxs)  
```

```python
"""problem-5: A store tracks units_sold (shape (n_days, n_products) ) and a unit_price vector (shape (n_products,) ). Compute the total
revenue per product across all days using matrix operations (no loop over days or products)."""


def total_revenue_by_product(units_sold, unit_price):
    """
    units_sold: 2D array, shape (n_days, n_products)
    unit_price: 1D array, shape (n_products,)
    returns: 1D array, shape (n_products,) — total revenue per product
    """
    return (units_sold * unit_price).sum(axis=0)

# Test
test_units = np.array([[10, 5], [8, 6], [12, 4]]) 
test_price = np.array([1000, 2000])
print(total_revenue_by_product(test_units, test_price))  
```

```python
"""problem-6: A grade book stores raw scores as a matrix scores of shape (n_students, n_assignments) , and a weights vector of shape
(n_assignments,) that sums to 1.0. Compute each student's final weighted score using matrix multiplication."""


def weighted_final_scores(scores, weights):
    """
    scores: 2D array, shape (n_students, n_assignments)
    weights: 1D array, shape (n_assignments,), sums to 1.0
    returns: 1D array, shape (n_students,) — each student's weighted final score
    """
    return scores @ weights

# Test
test_scores = np.array([[90, 80, 70], [60, 70, 80], [100, 90, 95]])
test_weights = np.array([0.5, 0.3, 0.2])
print(weighted_final_scores(test_scores, test_weights))  
```

```python
"""problem-7: A bakery sells two products. Product A uses 2 units of flour and 1 unit of sugar; Product B uses 1 unit of flour and 3 units of sugar. Selling
Product A costs 3,000 KRW and Product B costs 4,000 KRW in ingredients. Set up and solve the linear system to find the cost per unit of
flour and sugar."""


def solve_ingredient_costs():
    """
    returns: 1D array [cost_per_flour_unit, cost_per_sugar_unit]
    """

    A = np.array([[2, 1],
                  [1, 3]])
    b = np.array([3000, 4000])
    return np.linalg.solve(A, b)

print(solve_ingredient_costs())  # expected: [1000. 1000.]

```

```python
"""problem-8: Simulate flipping a fair coin 50,000 times ( 0 = tails, 1 = heads) and return the proportion of heads."""


def simulate_coin_flips(n_flips=50000, seed=42):
    """
    returns: float, proportion of flips that were heads (close to 0.5)
    """
    np.random.seed(seed)
    flips = np.random.randint(0, 2, size=n_flips)
    return flips.mean()

print(simulate_coin_flips())  
```

```python
"""problem-8: A call center's wait times are normally distributed with mean 5 minutes and standard deviation 1.5 minutes. Simulate 10,000 customer wait
times and return the percentage of customers who waited longer than 7 minutes."""


def pct_waiting_over_7min(n_customers=10000, seed=42):
    """
    returns: float, percentage (0-100) of customers who waited > 7 minutes
    """
    np.random.seed(seed)
    wait_times = np.random.normal(5, 1.5, size=n_customers)
    return (wait_times > 7).mean() * 100

print(pct_waiting_over_7min())  # expected: roughly 9-10 (percent)
```

```python

def simulate_portfolio_growth(start_value=1_000_000, n_days=252, seed=42):
    """returns float, final portfolio value after n_days of compound daily returns"""
    np.random.seed(seed)
    daily_returns = np.random.normal(0.0005, 0.015, size=n_days)
    growth_factors = 1 + daily_returns
    cumulative_growth = np.cumprod(growth_factors)
    return start_value * cumulative_growth[-1]

print(simulate_portfolio_growth())  # value should be reproducible with the given seed
```
